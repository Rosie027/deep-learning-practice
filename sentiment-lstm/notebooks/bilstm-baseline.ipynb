# 第二步：处理测试集

```python

from sklearn.model_selection import train_test_split

train_texts, val_texts, train_labels, val_labels = train_test_split(
    train_texts, train_labels, test_size=0.2, random_state=42, stratify=train_labels
)

print(f"新训练集：{len(train_texts)}，验证集：{len(val_texts)}")
```

### 第一步：数据预处理（文本清洗、分词、构建词表、转ID）

```python
# 定义了之前的清洗和分词函数
import re
import jieba
import numpy as np
from collections import Counter

def clean_text(text):
    # 去除特殊符号，保留中文、字母、数字
    text = re.sub(r'[^\u4e00-\u9fa5a-zA-Z0-9]', '', str(text))
    return text.strip()

def tokenize(text):
    text = clean_text(text)
    return jieba.lcut(text)

def build_vocab(texts, max_vocab=5000):
    all_words = []
    for text in texts:
        all_words.extend(tokenize(text))
    word_counts = Counter(all_words)
    vocab = {'<pad>': 0, '<unk>': 1}
    for word, _ in word_counts.most_common(max_vocab - 2):
        vocab[word] = len(vocab)
    return vocab

def text_to_ids(text, vocab, max_len=100):
    words = tokenize(text)
    ids = [vocab.get(w, vocab['<unk>']) for w in words[:max_len]]
    ids += [vocab['<pad>']] * (max_len - len(ids))
    return np.array(ids, dtype='int64')
```

```python
# 预处理转换

# 构建词表（仅用训练集）
vocab = build_vocab(train_texts, max_vocab=5000)
print(f"词表大小：{len(vocab)}")

# 文本转ID
max_len = 100
train_ids = np.array([text_to_ids(t, vocab, max_len) for t in train_texts])
val_ids   = np.array([text_to_ids(t, vocab, max_len) for t in val_texts])

train_labels = np.array(train_labels, dtype='int64')
val_labels   = np.array(val_labels, dtype='int64')

print(f"训练数据 shape: {train_ids.shape}")
print(f"验证数据 shape: {val_ids.shape}")
```

### 第2步：构建 DataLoader

```python
import paddle
from paddle.io import DataLoader, TensorDataset

batch_size = 64

train_dataset = TensorDataset([paddle.to_tensor(train_ids), paddle.to_tensor(train_labels)])
val_dataset   = TensorDataset([paddle.to_tensor(val_ids),   paddle.to_tensor(val_labels)])

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
```

### 第3步：定义模型

```python
import paddle.nn as nn

class SentimentLSTM(nn.Layer):
    def __init__(self, vocab_size, embed_dim=128, hidden_size=128, num_classes=2):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, hidden_size, 
                            num_layers=1, 
                            direction='bidirectional')
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(hidden_size * 2, num_classes)

    def forward(self, x):
        x = self.embedding(x)
        x, (h, c) = self.lstm(x)
        x = paddle.mean(x, axis=1)
        x = self.dropout(x)
        x = self.fc(x)
        return x

model = SentimentLSTM(vocab_size=len(vocab))

# 手动打印模型结构
print(model)
total_params = sum(p.numel() for p in model.parameters())
print(f"模型总参数量：{total_params}")
```

### 第4步：训练模型

```python
criterion = nn.CrossEntropyLoss()
optimizer = paddle.optimizer.Adam(learning_rate=0.001, parameters=model.parameters())

epochs = 5
train_losses = []
val_accs = []

for epoch in range(epochs):
    # 训练阶段
    model.train()
    total_loss = 0
    for batch_x, batch_y in train_loader:
        pred = model(batch_x)
        loss = criterion(pred, batch_y)
        loss.backward()
        optimizer.step()
        optimizer.clear_grad()
        total_loss += loss.item()          # 修改
    
    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    # 验证阶段
    model.eval()
    correct = 0
    total = 0
    with paddle.no_grad():
        for batch_x, batch_y in val_loader:
            pred = model(batch_x)
            pred_label = paddle.argmax(pred, axis=1)
            correct += (pred_label == batch_y).sum().item()  # ← 同理修改这里
            total += batch_y.shape[0]
    acc = correct / total
    val_accs.append(acc)
    
    print(f"Epoch {epoch+1}/{epochs} | Train Loss: {avg_loss:.4f} | Val Acc: {acc:.4f}")

print("训练完成！")
```

### 第5步：完整评估（用于实验报告）

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

model.eval()
all_preds = []
all_labels = []
with paddle.no_grad():
    for batch_x, batch_y in val_loader:
        pred = model(batch_x)
        pred_label = paddle.argmax(pred, axis=1)
        all_preds.extend(pred_label.numpy().tolist())
        all_labels.extend(batch_y.numpy().tolist())

# 计算各项指标
acc = accuracy_score(all_labels, all_preds)
p   = precision_score(all_labels, all_preds, average='binary')
r   = recall_score(all_labels, all_preds, average='binary')
f1  = f1_score(all_labels, all_preds, average='binary')

print("======= 验证集评估结果 =======")
print(f"准确率 Accuracy : {acc:.4f}")
print(f"精确率 Precision: {p:.4f}")
print(f"召回率 Recall   : {r:.4f}")
print(f"F1值 F1-score  : {f1:.4f}")
print("\n详细分类报告：")
print(classification_report(all_labels, all_preds, target_names=['负面', '正面']))
```
